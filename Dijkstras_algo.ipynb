```python
from queue import PriorityQueue
class Solution:
    def dijkstra(self, V, edges, S, E): # S is starting node, V is number of vertices, E is end node

            adj = [[] for _ in range(V)]

            path_length = [10**9] * V
            parents = [0]*V
            path_length[S] = 0
            parents[S] = -1

            for u, v, distance in edges:

                adj[u].append((v, distance))
                adj[v].append((u, distance))
            
            pq = PriorityQueue()

            pq.put((0, S))

            while not pq.empty():

                gap , node = pq.get()

                for nbr , path in adj[node]:

                    one_distance = path + gap

                    if one_distance < path_length[nbr]:
                        path_length[nbr] = one_distance
                        parents[nbr] = node

                        pq.put((one_distance, nbr))
            
            if path_length[E] == 10**9:
                return -1
            
            path = []
            node = E

            while node != -1:
                path.append(node)
                node = parents[node]

            path = path[::-1]


            return path


            


```

```python
if __name__ == '__main__':

    sol = Solution()

    V = 5
    edges = [
        (0, 1, 2),
        (0, 2, 4),
        (1, 3, 1),
        (2, 3, 3),
        (3, 4, 5)
            ]
    S = 0
    E = 4

    ans = sol.dijkstra(V, edges, S, E)

    print(ans)
```
