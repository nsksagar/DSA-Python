
### Storing in a matrix way (adjacency matrix)


Number of cells in the matrix = (N+1)x(N+1) where N is number of nodes

Space used is N^2

```python
def adjacency_matrix(nodes, edges_list):

    adj = [[0 for _ in range(nodes+1)] for _ in range(nodes+1)]

    for u, v in edges_list:

        adj[u][v] = 1
        adj[v][u] = 1

    return adj 
```

### Storing in list way


1 based indexing


space used is 2N 

```python
def adjacency_list(nodes, edges_list):

    adj_list = [[] for _ in range(nodes)]


    for u , v in edges_list:

        adj_list[u].append(v)

        adj_list[u].append(u)

    return adj_list
```

### How do you store the weighted graph?


#### Matrix method 
1. In adjacency matrix instead of just 1s you will write the weight if the edges
2. In adjacency list instead of storing just single node values, you will store pairs (node value , edge weight)
