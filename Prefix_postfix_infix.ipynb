```python
def infixtoprefix(s):

    s = reverse(s)
    ans = ''
    stack = []

    for i in s:

        if i.isalnum():
            ans += i 
        elif i == '(':
            stack.append(i)
        elif i == ')':
            while len(stack) != 0 and stack[-1] != '(':
                ans += stack.pop()
            stack.pop()
        else: 

            if i == '^':

                while len(stack) != 0 and priority(i) <= priority(stack[-1]):
                    ans += stack.pop()
            else:
                while len(stack) != 0 and priority(i) < priority(stack[-1]):
                    ans += stack.pop()
            stack.append(i)
                
            
    while len(stack) != 0 : 
        ans += stack.pop()
    
    ans = reverse(ans)
    return ans

def priority(x):

    if x == '^':
        return 3
    elif x == '*' or x == '/':
        return 2 
    elif x == '+' or x == '-':
        return 1
    else:
        return -1

def reverse(x):

    x = list(x)
    left = 0 
    right = len(x)-1

    while left <= right:

        if left != right:
            if x[left] == '(':
                x[left] = ')'
            elif x[left] == ')':
                x[left] = '('
    
            if x[right] == '(':
                x[right] = ')'
            elif x[right] == ')':
                x[right] = '('     
        else: 
            if x[left] == '(':
                x[left] = ')'
            elif x[left] == ')':
                x[left] = '('

        x[left], x[right] = x[right], x[left]

        left += 1
        right -=1 

    x = ''.join(x)

    return x
```
