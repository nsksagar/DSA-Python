```python
def convert2binary(n):

    if n == 0:
        return 0
    out = ''
    while n > 0:

        out = str(n % 2) + out
        n = n//2 

    return int(out)

print(convert2binary(13))
```

```python
def convertbin2num(n):

    n = str(n)
    out = 0

    for i in range(len(n)-1, -1, -1):

        out += int(n[i]) * (2**(len(n)-i-1))

    return out

print(convertbin2num(111))
```

```python
def onescomplement(n):

    bin = str(convert2binary(n))

    out = ''

    for i in bin:

        if i == '1':
            out += '0'
        else:
            out += '1'

    return out

print(onescomplement(13))
```

```python
def twoscomplement(n):

    onc = onescomplement(n)

    out = ''
    carry = '1'

    for i in range(len(onc)-1, -1, -1):

        if onc[i] == '1' and carry == '1':
            out = '0' + out
            carry = '1'
        elif onc[i] == '1' and carry == '0':
            out = '1' + out 
            carry = '0'
        elif onc[i] == '0':
            if carry == '1':
                out = '1' + out 
                carry = '0'
            else:
                out = '0' + out 
                carry = '0'
    if carry == '1':
        out = carry + out 
    return out

print(twoscomplement(13))
```

```python
def ithbitset(n, i):

    return (n >> i) & 1

print(ithbitset(13, 1))
```

```python
def swap(a,b):

    a = a ^ b
    b = a ^ b
    a = a ^ b
    return a,b

print(swap(5,6))
    
```

```python
def setithbit(n, i):

    return n | (1 << i)

print(setithbit(9, 2))
```

```python
def countsetbits(n):

    count = 0 
    while n > 0:

        if n & 1 == 1:
            count +=1 
        n >>= 1

    return count
```
