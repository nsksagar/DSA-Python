```python
class Node:
    def __init__(self, val):
        self.val = val 
        self.next = None 

```

```python
def convert(array):

    if not array:
        return None

    head = Node(int(array[0]))

    curr = head

    for i in range(1,len(array)):

        

        curr.next = Node(int(array[i]))

        curr = curr.next 

    return head 
```

```python
head = convert([1,2,3,4,5])
```

```python
def convert_LL_list(head): 

    if not head: 
        return None 
    

    arr = []

    curr = head

    while curr:

        arr.append(curr.val)


        curr = curr.next 

    return arr
```

```python

print(convert_LL_list(head))
```

```python
def addOne(head):


    if not head: 
        return None 
    

    lst = convert_LL_list(head)

    def convert_list_int(lst):

        res = 0 

        for i in lst:

            res = 10*res + i

        return res 
    
    out = convert_list_int(lst) + 1 

    out = str(out)


    array = list(out)

    head = convert(array)


    return head





    


```

```python
array = [9,9,9]

head = convert(array)

new_head = addOne(head)

convert_LL_list(new_head)
```
