```python
def recc(name, N):
    if N == 0:
        return 

    print(name)
    recc(name, N-1)

recc('sagar', 5)
```

```python
def continueNum(N):

    if N == 0:
        return 
    continueNum(N-1)
    print(N)
continueNum(10)

    
```

```python
def reverseNum(N):

    if N == 0:
        return 
    print(N)
    reverseNum(N-1)

reverseNum(10)
```

```python
def cntNum(N, count = 1):
    if count > N:
        return 

    print(count)

    cntNum(N , count + 1)

cntNum(10)
```

```python
def sum(N):

    if N <= 0: 
        return 0
    return N + sum(N-1)

print(sum(2))
```

```python
def fact(N):

    if N <= 0:
        return 1

    return N * fact(N-1)

print(fact(5))
```

```python
def fibb(N):

    if N == 0:
        return 0
    elif N == 1:
        return 1 

    return fibb(N-1) + fibb(N-2)

fibb(7)
```

```python
def reverse(array, left, right):

    if left >= right:
        return array

    array[left], array[right] = array[right], array[left]
    return reverse(array, left+1, right-1)

    

print(reverse([1,2,3,4,5], 0, 4))
```

```python
def palindrome_check(string, left, right):
    if left >= right:
        return True

    if string[left] != string[right]:
        return False

    else:
        return palindrome_check(string, left +1 , right-1)

print(palindrome_check('sagar', 0, len('sagar')-1))
```
