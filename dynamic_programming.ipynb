```python

def fib(n):
    if n <= 1:
        return n 
    
    return fib(n-1) + fib(n-2)

```

```python
def fib_memo(n, memo = None):

    if memo is None:
        memo = {}

    if n <= 1:
        return n 
    
    if n in memo:
        return memo[n]
    
    memo[n]  = fib_memo(n-1, memo) + fib_memo(n-2, memo)
    return memo[n]

print(fib_memo(10))
```

```python
def fib_tab(n):

    dp = [0]* (n+1)

    dp[1] = 1

    for i in range(2, n+1):

        dp[i] = dp[i -1] + dp[i-2]

    return dp

print(fib_tab(20))
```

```python
def fib_memo_2(n , dp):

    if n <= 1:
        return n 
    
    if dp[n] != -1:
        return dp[n]
    
    dp[n] = fib_memo_2(n-1, dp) + fib_memo_2(n-2, dp)

    return dp[n]

n = 10
dp = [-1] * (n+1)

print(fib_memo_2(n, dp))
```

```python
def fib_space(n):

    prev = 1 
    prev_2 = 0 

    for i in range(2, n+1):

        out = prev + prev_2 

        prev_2 = prev
        prev = out
    
    return out 

print(fib_space(10))
        
```
